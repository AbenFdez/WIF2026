### 1. Importar librerías y paquetes

```python
#Importar librerías
import pandas as pd
```

### 2. Importar datos de origen

#### 2.1 Importar datos de Instalaciones

```python
#Importar datos de Instalaciones (salida del notebook 02)
df_instalaciones = pd.read_excel(
    "../Data/Gold/Dim/MaestroInstalacionesGalicia.xlsx"
)

# Renombrar columnas
df_instalaciones = df_instalaciones.rename(columns={
    "propietario": "Propietario",
    "Cod-prv":"Cod_prv"
})

# Visualizar los resultados
df_instalaciones.head(5)
```

#### 2.2 Importar datos de Ingresos

```python
# Importar datos de 2025
df_ingresos_2025 = pd.read_excel("../Data/Raw/Dummy/Generacion_y_Ventas/Ingresos_parques_eolicos_Galicia_2025.xlsx")

# Visualizar los resultados
df_ingresos_2025.head(5)
```

```python
# Importar datos de 2024
df_ingresos_2024 = pd.read_excel("../Data/Raw/Dummy/Generacion_y_Ventas/Ingresos_parques_eolicos_Galicia_2024.xlsx")

# Visualizar los resultados
df_ingresos_2024.head(5)
```

```python
# Importar datos de 2023
df_ingresos_2023 = pd.read_excel("../Data/Raw/Dummy/Generacion_y_Ventas/Ingresos_parques_eolicos_Galicia_2023.xlsx")

# Eliminar columnas que no necesitamos
df_ingresos_2023 = df_ingresos_2023.drop(columns=["base_2024_aux"])

# Visualizar los resultados
df_ingresos_2023.head(5)
```

### 3. Transformación de datos (tablas hechos)

#### 3.1 Tabla Hechos Ingresos Eólicas (general) 

```python
# Concatener dataframes
df_ingresos_eolicas = pd.concat(
    [df_ingresos_2025, df_ingresos_2024, df_ingresos_2023],
    ignore_index=True
)
#df_ingresos_eolicas.head(10)

# Renombrar columnas
df_ingresos_eolicas = df_ingresos_eolicas.rename(columns={
    "Año": "Ejercicio",
    "Clave_Registro":"Id_Instalacion",
    'energia_generada_gwh':'Energia_generada_gwh',
    'energia_vendida_gwh':'Energia_vendida_gwh', 
    'potencia_instalada_mw':'Potencia_instalada_mw',
    'precio_mercado_mwh_euro':'Precio_mercado_mwh_euro', 
    'ingresos_estimados_euro':'Ingresos_estimados_euro'
})

# Incluir Columna de Codigo_Mes
meses = {
    "Enero": 1,
    "Febrero": 2,
    "Marzo": 3,
    "Abril": 4,
    "Mayo": 5,
    "Junio": 6,
    "Julio": 7,
    "Agosto": 8,
    "Septiembre": 9,
    "Octubre": 10,
    "Noviembre": 11,
    "Diciembre": 12
}
df_ingresos_eolicas["Codigo_Mes"] = df_ingresos_eolicas["Mes"].map(meses)

# Incluir Columna de Codigo_Mes_str
df_ingresos_eolicas["Codigo_Mes_str"] = df_ingresos_eolicas["Mes"].map(meses).astype(str).str.zfill(2)

# Incluir Columna de Fecha
df_ingresos_eolicas["Fecha"] = pd.to_datetime(
    df_ingresos_eolicas["Ejercicio"].astype(str) + "-" + df_ingresos_eolicas["Codigo_Mes_str"] + "-01"
)

# Eliminar columnas que no se necesitan
df_ingresos_eolicas = df_ingresos_eolicas.drop(columns=["Codigo_Mes_str"])

# Alta columna "Id_Ingresos"
df_ingresos_eolicas['Id_Ingresos'] = df_ingresos_eolicas["Id_Instalacion"] + "_" +  df_ingresos_eolicas["Ejercicio"].astype(str) + "_" + df_ingresos_eolicas["Codigo_Mes"].astype(str)

# Reordenar columnas
df_ingresos_eolicas = df_ingresos_eolicas[['Ejercicio', 'Codigo_Mes', 'Mes', 'Fecha', 'Id_Ingresos', 'Id_Instalacion', 
                                           'Nombre_parque', 'Energia_generada_gwh','Energia_vendida_gwh',
                                           'Potencia_instalada_mw','Precio_mercado_mwh_euro', 'Ingresos_estimados_euro']]
# Visualizar los resultados
df_ingresos_eolicas
```

```python
#Join para incluir información adicional de aerogneradores, tecnología y otros
df_ingresos_eolicas_vf = df_ingresos_eolicas.merge(
    df_instalaciones[
        ["Id_Instalacion",'Id_Propietario', 'Propietario','Cod_prv', 'Provincia']
    ],
    how="left",
    left_on="Id_Instalacion",
    right_on="Id_Instalacion"
)

#Excluir registros con valores en blanco o nulos en el campo Id_Propietario
df_ingresos_eolicas_vf = df_ingresos_eolicas_vf[
    df_ingresos_eolicas_vf["Id_Propietario"].notna() &
    (df_ingresos_eolicas_vf["Id_Propietario"].str.strip() != "")
]

#Reordenar columnas
df_ingresos_eolicas_vf = df_ingresos_eolicas_vf[['Ejercicio', 'Codigo_Mes', 'Mes', 'Fecha', 'Id_Ingresos', 'Id_Instalacion', 
                                                 'Nombre_parque', 'Id_Propietario', 'Propietario','Cod_prv', 'Provincia',
                                                 'Energia_generada_gwh','Energia_vendida_gwh', 'Potencia_instalada_mw',
                                                 'Precio_mercado_mwh_euro', 'Ingresos_estimados_euro']]

#Visualizar los resultados
df_ingresos_eolicas_vf
```

#### 3.2 Tabla Hechos Generación 

```python
#DF para tabla de Generación
df_generacion = df_ingresos_eolicas_vf[['Ejercicio', 'Codigo_Mes', 'Mes', 'Fecha', 'Id_Ingresos', 'Id_Instalacion', 
                                        'Nombre_parque', 'Id_Propietario', 'Propietario','Cod_prv', 'Provincia',
                                        'Energia_generada_gwh']]

# Renombrar columnas
df_generacion = df_generacion.rename(columns={
    "Id_Ingresos": "Id_Generacion"
})

#Visualizar los resultados
df_generacion.head(10)
```

#### 3.3 Tabla Hechos Ventas

```python
#DF para tabla de Ventas
df_ventas = df_ingresos_eolicas_vf[['Ejercicio', 'Codigo_Mes', 'Mes', 'Fecha', 'Id_Ingresos', 'Id_Instalacion', 
                                   'Nombre_parque', 'Id_Propietario', 'Propietario','Cod_prv', 'Provincia',
                                    'Energia_vendida_gwh', 'Potencia_instalada_mw','Precio_mercado_mwh_euro', 
                                    'Ingresos_estimados_euro']]
# Renombrar columnas
df_ventas = df_ventas.rename(columns={
    "Id_Ingresos": "Id_Ventas"
})

#Visualizar los resultados
df_ventas.head(5)
```
