### 1. Importar librerías y paquetes

```python
# Importar librerias
import pandas as pd
import unicodedata
import os
```

### 2. Importar datos de origen

#### 2.1 Importar datos de Instalaciones de Galicia

```python
# Importar datos de instalaciones de Galicia
df_instalaciones_galicia = pd.read_excel("../Data/Raw/Public/MITECO/Galicia_InformeInstalaciones_04092026.xlsx")

# Visualizar los resultados
df_instalaciones_galicia
```

#### 2.2 Importar datos de Provincias

```python
# Importar datos de Provincias (del INE)
df_provincias = pd.read_excel(
    "../Data/Silver/D_provincias_ine.xlsx"
)

df_provincias = df_provincias[['Id_provincia', 'Id_ccaa', 'Cod-prv', 'Nombre_provincia',
       'Nombre_provincia_st', 'Nombre_provincia_completo']]

# Visualizar los resultados
df_provincias
```

#### 2.3 Importar datos de Propietarios

```python
# Importar datos de propietarios de instalaciones (del INEGA)
df_propietarios_instalaciones = pd.read_excel(
    "../Data/Raw/Public/INEGA/INEGA_Centrales_Eolicas_Galicia_202606.xlsx"
)

# Visualizar los resultados
df_propietarios_instalaciones.head(5)
```

#### 2.4 Importar datos de Tecnologia y Aerogeneradores

```python
# Importar datos de información de tecnologia y aerogeneradores (Varias fuentes)
df_instalaciones_infoaerogeneradors = pd.read_excel(
    "../Data/Raw/Dummy/Generadores_y_Tecnologos/MaestroInstalacionesGalicia_InfoAdicional.xlsx"
)

# Incluir campo Id_Instalacion
df_instalaciones_infoaerogeneradors['Id_Instalacion'] = df_instalaciones_infoaerogeneradors['Clave Registro'] + "-" + df_instalaciones_infoaerogeneradors['Número de fase'].astype("string")

# Visualizar los resultados
df_instalaciones_infoaerogeneradors.head(5)
```

### 3. Transformación de datos (Tabla D_Instalaciones)

#### 3.1 Inclusión de datos de Propietarios y Geografía

```python
# Añadir columna de CCAA
df_instalaciones_galicia['Id_ccaa'] = 12

# Renombrar columnas
df_instalaciones_galicia = df_instalaciones_galicia.rename(columns={
    "Provincia de la Instalación": "Provincia",
    "Municipio de la Instalación": "Municipio",
    "Potencia Instalada KW": "Potencia_kw",
    "Grupo Normativo": "Cod_Grupo",
    "Clave Registro": "Clave_Registro",
    "Nombre de Instalación": "Nombre_Instalacion",
    "Número de fase": "Fase",
    "Tipo de Inscripción": "Tipo_Inscripcion",
    "Fecha de Inscripción Definitiva": "Fecha_Inscripcion",
    "Número de Registro Autonómico Definitivo": "Num_Registro"
    
})

# Incluir campo Id_Instalacion
df_instalaciones_galicia['Id_Instalacion'] = df_instalaciones_galicia['Clave_Registro'] + "-" + df_instalaciones_galicia['Fase'].astype("string")

# Incluir campo Instalacion
df_instalaciones_galicia['Instalacion'] = df_instalaciones_galicia['Id_Instalacion'] + " " + df_instalaciones_galicia['Nombre_Instalacion']

# Filtrar registros de instalaciones eólicas (valores "b.2.1" y "b.2.2" en columna Cod_Grupo)
df_instalaciones_galicia = df_instalaciones_galicia[
    (df_instalaciones_galicia["Cod_Grupo"] == "b.2.1") |
    (df_instalaciones_galicia["Cod_Grupo"] == "b.2.2")
]

#Substutuir valores erróneos en la columna "Provincia"
df_instalaciones_galicia["Provincia"] = (
    df_instalaciones_galicia["Provincia"].replace({
        "La Coruña": "A Coruña",
        "Orense": "Ourense"
    })
)

#Substutuir valores erróneos en la columna "Num_Registro"
df_instalaciones_galicia["Num_Registro"] = (
    df_instalaciones_galicia["Num_Registro"].replace({
        "RE-04-24 y RE-05-85": "RE-05-85","RE-01-25":"RE-05-56",   "REP-05-45":"RE-07-97",
        "RP-26-08":"RE-10-29", "REP-09-09":"RE-09-09", "RE-05-18":"RE-05-08","REP-06-40":"RE-06-40", "REP-06-05":"RE-06-05"
    })
)

#Join para incluir id_provincia
df_instalaciones_galicia = df_instalaciones_galicia.merge(
    df_provincias[
        ["Nombre_provincia","Cod-prv"]
    ],
    how="left",
    left_on="Provincia",
    right_on="Nombre_provincia"
)

#Eliminar valores entre "()" en la columna Num_Registro
columnas = ["Num_Registro"]
for col in columnas:
    df_instalaciones_galicia[col] = (
        df_instalaciones_galicia[col]
        .str.replace(r"\s*\([^)]*\)", "", regex=True)
        .str.strip()
    )

#Join para incluir el propietario de la instalación (información de Inega)
df_instalaciones_galicia = df_instalaciones_galicia.merge(
    df_propietarios_instalaciones[
        ["Registro_inega","Id_Propietario","Propietario"]
    ],
    how="left",
    left_on="Num_Registro",
    right_on="Registro_inega"
)

# Excluir registros con valores en blanco o null en Id_Propietario
df_instalaciones_galicia = df_instalaciones_galicia[
    df_instalaciones_galicia["Id_Propietario"].notna() &
    (df_instalaciones_galicia["Id_Propietario"].str.strip() != "")
]

#Eliminar columnas que no se necesitan
df_instalaciones_galicia = df_instalaciones_galicia.drop(
    columns=["Nombre_provincia", "Registro_inega"]
)

#Reemplazar valores erróneos en la columna Fecha_Inscripcion
df_instalaciones_galicia["Fecha_Inscripcion"] = pd.to_datetime(
    df_instalaciones_galicia["Fecha_Inscripcion"].replace("N/D", pd.NA),
    errors="coerce"
)

#Reordenar columnas
df_instalaciones_galicia = df_instalaciones_galicia[['Id_Instalacion','Clave_Registro', 'Nombre_Instalacion', 'Instalacion',
                                                     'Id_Propietario','Propietario','Id_ccaa', 'Cod-prv','Provincia','Municipio',
                                                     'Num_Registro','Cod_Grupo', 'Fase',  'Potencia_kw',  'Tipo_Inscripcion',
                                                     'Fecha_Inscripcion']]

# Visualizar los resultados
df_instalaciones_galicia
```

#### 3.2 Inclusión de datos de Aerogeneradores y Tecnología

```python
#Join para incluir información adicional de aerogneradores, tecnología y otros
df_instalaciones_galicia_vf = df_instalaciones_galicia.merge(
    df_instalaciones_infoaerogeneradors[
        ["Id_Instalacion",'Num_aerogeneradores', 'Fabricante',
       'Modelo', 'Potencia_unitaria_kw']
    ],
    how="left",
    left_on="Id_Instalacion",
    right_on="Id_Instalacion"
)

#Reordenar columnas
df_instalaciones_galicia_vf = df_instalaciones_galicia_vf[['Id_Instalacion', 'Clave_Registro', 'Nombre_Instalacion',
       'Id_Propietario', 'Propietario', 'Id_ccaa', 'Cod-prv', 'Provincia',
       'Municipio', 'Num_Registro', 'Cod_Grupo','Potencia_kw', 'Fase', 'Num_aerogeneradores', 'Fabricante', 'Modelo',
       'Potencia_unitaria_kw','Tipo_Inscripcion', 'Fecha_Inscripcion']]

df_instalaciones_galicia_vf
```
